```python
import numpy as np
import math
```

```python
def busqueda_binaria(n_busca, flujo, n_intentos = 0):
    if len(flujo) != 0:
        min_ = flujo.min()
        max_ = flujo.max()
        promedio = math.floor((min_+max_)/2)
        if promedio == n_busca:
            print("se encontró en el intento: ", n_intentos)
        else:
            n_intentos += 1
            centinela = np.where(flujo == promedio)[0][0]
            if promedio < n_busca:
                flujo = flujo[centinela:]
            else:
                flujo = flujo[:centinela]
            print(promedio, min_, max_)
            busqueda_binaria(n_busca, flujo, n_intentos)
    else:
        print('No se encontró el elemento buscado')
```

```python
flujo = np.arange(1, 201)
```

```python
n_busca = 145
```

```python
busqueda_binaria(n_busca, flujo)
```

### Análisis de Resultados: Búsqueda Binaria

Tras corregir el error de tipado (Type Error) y buscar un valor numérico válido (ej. 145) dentro del arreglo de 200 elementos, el algoritmo ejecutó la búsqueda binaria de manera exitosa.

**Observaciones de la ejecución:**
La consola nos imprimió el rastro de la estrategia "Divide y Vencerás" (`Divide and Conquer`). En lugar de revisar los 200 números uno por uno (lo que habría tomado 145 intentos usando una búsqueda lineal), el algoritmo partió el arreglo a la mitad en cada paso, descartando el 50% de las opciones restantes en cada iteración.

* **Eficiencia (Complejidad de Tiempo):** Encontrar el número tomó únicamente **5 intentos**. Esto demuestra de forma práctica que la complejidad temporal de la Búsqueda Binaria es logarítmica: $\mathcal{O}(\log n)$. Incluso si nuestro arreglo tuviera 1 millón de datos, el algoritmo lo habría encontrado en apenas unos 20 intentos.

# BÚSQUEDA BINARIA PARA TEXTO (NOMBRES)
# ==========================================

```python
def busqueda_binaria_nombres(nombre_busca, lista, n_intentos=1):
    # VALIDACIÓN 1: Si la lista se queda vacía, el nombre no existe.
    if len(lista) == 0:
        print(f"No se encontró '{nombre_busca}' en la lista.")
        return

    # VALIDACIÓN 2: Estandarizar a minúsculas para una comparación justa.
    nombre_busca_min = nombre_busca.lower()

    # Calculamos el ÍNDICE de la mitad (división entera)
    mitad_idx = len(lista) // 2
    elemento_medio = lista[mitad_idx]

    print(f"Intento {n_intentos}: Buscando en {lista} -> Revisando a '{elemento_medio}'")

    # Comparamos ignorando mayúsculas
    if elemento_medio.lower() == nombre_busca_min:
        print(f"\n¡Éxito! Se encontró '{elemento_medio}' en el intento {n_intentos}.")
    else:
        n_intentos += 1

        # VALIDACIÓN 3: Comparación alfabética real (A es menor que Z)
        if elemento_medio.lower() > nombre_busca_min:
            # Si el medio es alfabéticamente mayor, cortamos y nos quedamos la mitad izquierda
            nueva_lista = lista[:mitad_idx]
        else:
            # Si es menor, nos quedamos la mitad derecha (saltando el elemento que ya revisamos)
            nueva_lista = lista[mitad_idx + 1:]

        # Llamada recursiva con la lista recortada
        busqueda_binaria_nombres(nombre_busca, nueva_lista, n_intentos)
```

```python
nombres_originales = ['Leo', 'Sebastián', 'Arturo', 'Rafael', 'Bruno', 'Luis','Jesus']
```

```python
# VALIDACIÓN 4: ¡ORDENAMIENTO! La búsqueda binaria JAMÁS funciona en listas desordenadas.
nombres_ordenados = sorted(nombres_originales)

print("Lista original:", nombres_originales)
print("Lista ordenada para buscar:", nombres_ordenados)
print("-" * 50)
```

```python
# Buscamos a Leo (puedes probar escribiéndolo todo en minúsculas: 'leo')
busqueda_binaria_nombres('leo', nombres_ordenados)
```

### Análisis: Implementación Correcta de Búsqueda Binaria (Strings)

Para hacer que el algoritmo funcione con texto, se reestructuró la lógica original y se agregaron **4 validaciones críticas** que el código anterior omitía:

**1. Ordenamiento Obligatorio (`sorted()`):**
El algoritmo de búsqueda binaria asume que si un elemento no es el que buscas, puedes descartar la mitad de la lista sabiendo que los demás son mayores o menores. Si la lista `['Leo', 'Sebastián', 'Arturo'...]` no está ordenada alfabéticamente desde el principio, el algoritmo descartaría nombres incorrectamente y fallaría. Por eso aplicamos `sorted(nombres_originales)` antes de invocar la función.

**2. Manejo Sensible a Mayúsculas (`.lower()`):**
En Python, la letra `"L"` mayúscula y la `"l"` minúscula tienen valores diferentes en la tabla ASCII. Si el usuario busca `"leo"` pero en la lista está `"Leo"`, el programa original diría que no son iguales. Al aplicar `.lower()` a ambos lados durante la comparación, blindamos el código contra errores de dedo.

**3. Partición por Índices (`len(lista) // 2`):**
El código anterior promediaba los *valores* del arreglo `(min + max) / 2`. Esto es imposible con cadenas de texto. La solución real para cualquier búsqueda binaria es ubicar el índice central del arreglo dividiendo su longitud entre 2 (división entera `//`).

**4. Prevención de Bucles Infinitos (`mitad_idx + 1`):**
Al hacer el recorte (slicing) del arreglo hacia la derecha, usamos `lista[mitad_idx + 1:]`. Al sumar ese `+1`, nos aseguramos de no volver a incluir en la nueva lista el elemento central que ya acabamos de evaluar, garantizando que el arreglo se reduzca a `0` si el elemento no existe, disparando nuestro caso base (`if len(lista) == 0`).

**Conclusión de la ejecución:**
Al buscar a `'Leo'`, el algoritmo partió el arreglo alfabético, ubicó la mitad, comparó lexicográficamente los *strings* (sabiendo que la L está después de la B pero antes de la R) y encontró el objetivo reduciendo drásticamente el número de comparaciones.

# EJERCICIO: BÚSQUEDA DE TEXTO EN PYSPARK (COLAB)
# ==========================================

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, lower
```

```python
# 1. Inicializamos la sesión de Spark (¡Aquí ya no se va a trabar!)
print("Iniciando motor de PySpark...")
spark = SparkSession.builder.appName("BusquedaNombres").getOrCreate()
```

```python
# 2. Creamos nuestro DataFrame distribuido con los nombres
nombres_originales = [('Leo',), ('Sebastián',), ('Arturo',), ('Rafaelo',), ('Bruno',), ('Luis',)]
df_nombres = spark.createDataFrame(nombres_originales, ["nombre"])

print("\nDataset distribuido en el clúster:")
df_nombres.show()
```

```python
# 3. Definimos el nombre a buscar (blindado contra errores de dedo)
nombre_busca = "leo"

print("-" * 50)
print(f"Buscando a '{nombre_busca}' de forma distribuida...")
```

```python
# 4. LA FORMA PYSPARK DE BUSCAR (Filtro distribuido)
resultado = df_nombres.filter(lower(col("nombre")) == nombre_busca.lower())
```

```python
# 5. Verificamos si existe en el clúster
if resultado.count() > 0:
    print(f"\n¡Éxito! Se encontró a '{nombre_busca}'.")
    resultado.show()
else:
    print(f"\nNo se encontró a '{nombre_busca}' en el dataset.")
```

### Análisis: Búsqueda de Datos en Computación Distribuida (PySpark)

En sistemas distribuidos (Big Data), el acceso aleatorio por índice de la Búsqueda Binaria clásica es un anti-patrón. En lugar de ordenar globalmente y buscar un índice central (lo cual causaría un *Shuffle* masivo en la red), en PySpark utilizamos `.filter()`.

Bajo el capó, el motor **Catalyst Optimizer** de Spark toma esta instrucción y la envía a todos los nodos (Workers) del clúster simultáneamente. Esto permite que la búsqueda se realice de forma paralela en la memoria de múltiples computadoras, encontrando el resultado sin necesidad de concentrar los datos en un solo lugar.
